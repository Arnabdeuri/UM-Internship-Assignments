```python
# Task 2 - Visualise the distributions of ratings and compensations:

# We will now visualise the distributions of employee ratings and
# compensations.
# We will create the following plots:

# *   A scatter plot of employee age and employee salary, healthy_eating and active_lifestyle, health_eating and salary
# *   A count plot of groups
# *   A histogram of employee salary
```

```python
# Creating scatter plot of the ‘Age’ and ‘Employee Salary’ features

plt.figure(figsize = (10,5))
sns.scatterplot(data = data ,x = 'age', y = 'salary', color = 'red', edgecolor= 'linen', alpha=0.5)
plt.title('Scatter Plot of Employee Age and Employee Salary')
plt.xlabel('Employee Age')
plt.ylabel('Employee Salary')
plt.show()
```

```python
# Observations:

# We observe that in general, there is a relationship between age and salary of the employee.
# As age increases, the salary also increases, though there are some outliers present as well.
```

```python
# Creating scatter plot of 'healthy_eating' and 'salary' features.

plt.figure(figsize = (10,5))
sns.scatterplot(data = data ,x = 'healthy_eating', y = 'salary', color = 'm', edgecolor= 'linen', alpha=0.5)
plt.title('Scatter Plot of Healthy Eating and Salary')
plt.xlabel('Healthy Eating')
plt.ylabel('Salary')
plt.show()
```

```python
# Observations:

# We observe that in general, there is a relationship between healthy_eating and salary of the employee.
# As healthy_eating increases, salary also increases.
```

```python
# Creating countplot of 'groups' feature.

plt.figure(figsize = (10,5))
sns.countplot(data = data ,x = 'groups',hue ='groups',edgecolor= 'Ivory', alpha=0.7)     # 'hue' separates data into categories using different colors.

plt.title('Count Plot of Groups')
plt.xlabel('Groups')
plt.ylabel('Count')
plt.show()
```

```python
# Observations:

# We see that most employees either belong to Group A or Group O, with Group A having maximum frequency.
```

```python
# Creating histogram for 'salary' feature.

plt.figure(figsize = (11,6))
sns.histplot(data = data ,x = 'salary', color='blue', edgecolor= 'linen', alpha=0.7, bins=10)
plt.title('Histogram of Salary')
plt.xlabel('Salary')
plt.ylabel('Count')
plt.show()
```

```python
# Observations:

# We see that the salaries are not uniform, which is not necessarily a discripency.
# But we observe that there are employees with salaries on either extrimities of the histogram.
# Based on the histogram, we can see that the count of highest earning employee is very less
```

```python
# Observations from Task 2:

# 1. Generally, as the employee's age increase, there is an increament in the employee's salary.
# 2. Generally, as the employee's healthy_eating habit increase, there is an increament in the employee salary as well.
# 3. However, the employee salary values are more spread out.
# 4. An employee of healthy_eating having greater than 8 can be taken as 8 only.
# 5. An employee having salary less 1000 can be considered as a lower salary.
```

```python
# Task 3 - Subset the data based on thresholds
# We will now subset the original data frame based on the following conditions:

# • Employees with healthy_eating greater than 8
# • Employees with Salary less than 1000
# • Employees with healthy_eating greater than 8 and with salary less than 1000
```

```python
#  Employees with healthy_eating greater than 8.

sub1=data[data['healthy_eating']>8]
sub1
```

```python
#  Employees having Salary less than 1000.

sub2=data[data['salary']<1000]
sub2
```

```python
#  Employees with healthy_eating greater than 8 and with salary less than 1000.

sub=data[(data['healthy_eating']>8) & (data['salary']<1000)]
sub
```

```python
# Observations:

# 1. The only employee seemingly facing a discrepancy in salary as compared to healthy_eating is employee who is having id = 26
# 2. The employee with id having 26 has a salary of 700
```
